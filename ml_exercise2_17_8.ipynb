```python
import pandas as pd

d = pd.DataFrame({
    'A': [1, 2, 3],
    'B': [4, 5, 6]
})

d2 = pd.DataFrame({
    'A': [7, 8, 9],
    'B': [10, 11, 12]
})

d = pd.concat([d, d2])

print(d)
```
